```python
COUNSELOR_RULES = {
    "answer_source": (
        "Answer course-related questions using only verified "
        "information from the institute's knowledge base."
    ),

    "fees_and_discounts": (
        "Never invent fees, discounts, offers or additional charges."
    ),

    "placements": (
        "Never promise a job or guaranteed placement unless "
        "explicit institute-approved information supports the claim."
    ),

    "missing_information": (
        "If information is unavailable or unclear, say you cannot "
        "verify it and offer to connect the student with a counselor."
    ),

    "student_profiling": (
        "Understand the student's education, skills, goals, budget "
        "and schedule through natural conversation."
    ),

    "recommendations": (
        "Recommend courses based on the student's needs and explain "
        "the reason honestly. Never guarantee outcomes."
    ),

    "language": (
        "Reply in the student's language and conversational style. "
        "Support English, Hindi, Marathi and mixed-language messages."
    ),

    "human_handover": (
        "Refer unresolved questions to a human counselor when needed."
    )
}

print("Counselor rules created:", len(COUNSELOR_RULES))

for rule, instruction in COUNSELOR_RULES.items():
    print(f"- {rule}: {instruction}")
```

```python
required_fields = {
    "course_id",
    "name",
    "description",
    "fee_inr",
    "duration",
    "schedule",
    "eligibility",
    "syllabus"
}

course_ids = set()
errors = []

for course in INSTITUTE["courses"]:

    # Check that every course has the required information
    missing = required_fields - course.keys()

    if missing:
        errors.append(
            f"{course.get('name', 'Unknown course')}: "
            f"Missing fields {sorted(missing)}"
        )

    # Check for duplicate course IDs
    course_id = course.get("course_id")

    if course_id in course_ids:
        errors.append(f"Duplicate course ID: {course_id}")

    course_ids.add(course_id)

    # Check that the fee is a valid non-negative number
    fee = course.get("fee_inr")

    if not isinstance(fee, (int, float)) or isinstance(fee, bool) or fee < 0:
        errors.append(
            f"Invalid fee for course: {course.get('name', 'Unknown')}"
        )

if errors:
    print("Data validation failed:")

    for error in errors:
        print("-", error)

else:
    print("All course data checks passed.")
    print("Courses validated:", len(INSTITUTE["courses"]))
```

```python
def find_courses(query: str) -> list[dict]:
    """
    Search course information using simple keyword matching.
    This is our temporary retrieval method before implementing RAG.
    """
    query = query.strip().casefold()

    if not query:
        return []

    query_words = set(query.split())
    matches = []

    for course in INSTITUTE["courses"]:
        searchable_text = " ".join([
            course["name"],
            course["description"],
            course["duration"],
            course["schedule"],
            course["eligibility"],
            " ".join(course["syllabus"])
        ]).casefold()

        # Match if the query phrase or any meaningful query word appears
        if query in searchable_text or any(
            len(word) > 2 and word in searchable_text
            for word in query_words
        ):
            matches.append(course)

    return matches


# Test the search function
results = find_courses("Python")

if results:
    for course in results:
        print("Course:", course["name"])
        print(f"Fee: ₹{course['fee_inr']:,}")
        print("Duration:", course["duration"])
        print("Schedule:", course["schedule"])
        print("-" * 30)
else:
    print("No matching courses found.")
```
